# Asynchronous Programming - Practice
This notebook covers concurrency concepts, writing async coroutines using async/await, and running tasks concurrently using `asyncio.gather`.

> **Note:** In Jupyter notebooks, an event loop is already running, so you can call `await` directly at the top level instead of calling `asyncio.run()`.

### 1. Basic Async Coroutine
Declare a coroutine using `async def`, simulate non-blocking sleep with `await asyncio.sleep()`, and execute it using top-level `await`.

```python
import asyncio

async def fetch_data():
    print("Start fetching data...")
    # Simulate a network delay (non-blocking sleep)
    await asyncio.sleep(2)
    print("Data fetched!")
    return {"data": 123}

# In Jupyter, we can directly await the coroutine
result = await fetch_data()
print(result)
```

### 2. Concurrent Task Execution using asyncio.gather
Run multiple coroutines concurrently to drastically reduce execution time for I/O-bound operations.

```python
import asyncio
import time

async def call_api(service_name: str, delay: int):
    print(f"Calling {service_name}...")
    await asyncio.sleep(delay)
    print(f"{service_name} done!")
    return f"{service_name} response"

start_time = time.time()

# Run three API calls concurrently
results = await asyncio.gather(
    call_api("Auth Service", 2),
    call_api("Product Catalog", 1),
    call_api("Payment Gateway", 3)
)

end_time = time.time()
print(f"Results: {results}")
print(f"Total elapsed time: {end_time - start_time:.2f} seconds")
```
