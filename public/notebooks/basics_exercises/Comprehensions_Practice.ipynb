### 1. Lists: CRUD Operations and Sorting

Lists are ordered, mutable sequences of elements.

#### Create (C) & Read (R) - Indexing & Slicing

````python
languages = ["Python", "JavaScript"]
print("Languages:", languages)

fruits = ["Apple", "Banana", "Cherry", "Date", "Elderberry"]

# Indexing
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])

# Slicing
print("Slice [1:4]:", fruits[1:4])
print("Slice [:3]:", fruits[:3])
print("Slice [2:]:", fruits[2:])
print("Slice [::2]:", fruits[::2])
print("Reversed list [::-1]:", fruits[::-1])
````

#### Update (U)

````python
colors = ["red", "green"]
colors[1] = "blue"  # In-place modify
colors.append("yellow")  # Append
colors.insert(1, "orange")  # Insert at index
colors.extend(["purple", "pink"])  # Extend
print("Updated colors:", colors)
````

#### Delete (D)

````python
tasks = ["code", "test", "deploy", "debug"]
tasks.remove("test")  # Remove by value
popped = tasks.pop(1)  # Remove by index
print("Popped item:", popped)
del tasks[0]  # Delete via index
print("Remaining tasks:", tasks)
````

#### Sorting

````python
nums = [42, 7, 19, 88, 3]
print("Sorted Copy:", sorted(nums))

nums.sort()  # In-place
print("In-place sorted:", nums)
nums.sort(reverse=True)
print("Descending sorted:", nums)
````

### 2. Tuples: Operations, Indexing, and Slicing

Tuples are ordered, immutable sequences of elements. They support the exact same indexing and slicing syntax as lists.

#### Create (C) & Read (R)

````python
tup = (10, 20, 30, 40, 50)
single_tup = (5,)  # Note the trailing comma
print("Single element tuple:", single_tup)
print("First element:", tup[0])
print("Slice [1:4]:", tup[1:4])
print("Reversed:", tup[::-1])
````

#### Tuple Operations & Sorting

````python
t1 = (1, 2)
t2 = (3, 4)
print("Concatenation:", t1 + t2)
print("Repetition:", t1 * 3)

chars = ("a", "b", "a")
print("Count of 'a':", chars.count("a"))

data = (42, 7, 19)
print("Sorted list from tuple:", sorted(data))
print("Sorted tuple:", tuple(sorted(data)))
````

### 3. Dictionaries: CRUD Operations and Sorting

Dictionaries are key-value mappings where keys are unique and immutable.

#### Create (C) & Read (R)

````python
student = {"name": "Alice", "age": 25}
print("Name:", student["name"])
print("Safe get age:", student.get("age"))
print("Safe get location:", student.get("location", "Not Found"))
````

#### Update (U) & Delete (D)

````python
car = {"brand": "Tesla", "model": "Model 3"}
car["year"] = 2021  # Add
car["model"] = "Model Y"  # Update
car.update({"color": "red", "battery": "Long Range"})
print("Updated car:", car)

car.pop("color")  # Remove by key
print("After pop:", car)
````

#### Sorting Dictionaries

````python
scores = {"Charlie": 90, "Alice": 95, "Bob": 85}
print("Sorted by keys:", dict(sorted(scores.items())))
print("Sorted by values:", dict(sorted(scores.items(), key=lambda item: item[1])))
````

### 4. Sets: CRUD Operations and Sorting

Sets are unordered collections of unique, immutable elements.

#### Create (C) & Read (R)

````python
s = {1, 2, 3, 3}  # Unique values only
print("Set:", s)
print("2 in set:", 2 in s)
````

#### Update (U) & Delete (D)

````python
countries = {"India", "USA"}
countries.add("UK")
countries.update(["Canada", "Germany"])
print("Set after updates:", countries)

countries.discard("Reebok")  # Safe delete
countries.remove("USA")  # Throws error if missing
print("Final set:", countries)
````

#### Sorting Sets

````python
values = {42, 3, 17, 8}
print("Sorted set (returns list):", sorted(values))
````

### 5. Queues: Using `collections.deque`

A queue is a First-In-First-Out (FIFO) collection. `collections.deque` implements this with fast $O(1)$ operations on both ends.

````python
from collections import deque

queue = deque()
queue.append("Alice")  # Enqueue
queue.append("Bob")

print("Dequeued:", queue.popleft())  # Dequeue FIFO
print("Remaining queue:", queue)
````

### 6. Why Comprehensions?

Comprehensions allow generating new collections from existing iterables using a concise syntax.

````python
# Traditional loop
squares = []
for number in range(1, 6):
    squares.append(number ** 2)
print("Loop squares:", squares)

# Comprehension
squares_comp = [number ** 2 for number in range(1, 6)]
print("Comprehension squares:", squares_comp)
````

### 7. List Comprehensions

````python
numbers = range(1, 11)
evens = [num for num in numbers if num % 2 == 0]
print("Evens:", evens)

labels = ["Even" if num % 2 == 0 else "Odd" for num in range(1, 6)]
print("Labels:", labels)

matrix = [[1, 2], [3, 4]]
flat = [num for row in matrix for num in row]
print("Flattened:", flat)
````

### 8. Dictionary Comprehensions

````python
numbers = [1, 2, 3]
squares = {num: num ** 2 for num in numbers}
print("Squares dictionary:", squares)
````
