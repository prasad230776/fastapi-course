# Working with Data - Practice
This notebook covers plain text file handling, JSON serialization/deserialization, and CSV reading/writing.

### 1. Reading Text Files (Entire Content)
Open and read an entire text file at once using the `with open()` statement.

```python
# Create sample file first
with open("sample.txt", "w") as file:
    file.write("Line 1\nLine 2\nLine 3\n")

# Read entire content
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)
```

### 2. Reading Text Files (Line-by-Line)
Iterate over a file object to read it line-by-line, which is highly memory-efficient.

```python
with open("sample.txt", "r") as file:
    for line in file:
        print(line.strip())
```

### 3. Writing and Appending Text Files
Use the `'w'` flag to overwrite a file and `'a'` to append text to the end of the file.

```python
# Overwrite write
with open("output.txt", "w") as file:
    file.write("Hello, World!\n")
```

```python
# Append
with open("output.txt", "a") as file:
    file.write("Appending a new line!\n")

with open("output.txt", "r") as file:
    print(file.read())
```

### 4. JSON Serialization (dumps)
Convert a Python dictionary or list into a JSON string format.

```python
import json
user_profile = {
    "name": "Alice",
    "age": 25,
    "skills": ["Python", "Machine Learning"],
    "is_active": True
}
json_string = json.dumps(user_profile, indent=4)
print(json_string)
```

### 5. JSON Deserialization (loads)
Parse a JSON string representation back into a native Python dictionary/list.

```python
import json
json_string = '{"name": "Alice", "age": 25, "is_active": true}'
data_dict = json.loads(json_string)
print(data_dict)
print(type(data_dict))
```

### 6. Writing and Reading JSON Files (dump & load)
Save structured Python data directly to a JSON file and reload it.

```python
import json
user_profile = {"name": "Alice", "age": 25, "skills": ["Python"]}

# Write
with open("profile.json", "w") as file:
    json.dump(user_profile, file, indent=4)
```

```python
# Read
with open("profile.json", "r") as file:
    loaded_data = json.load(file)
    print("Loaded skills:", loaded_data["skills"])
```

### 7. Reading CSV Files as Lists
Read CSV rows sequentially as list values using the standard `csv.reader`.

```python
import csv
# Setup employee CSV
with open("employees.csv", "w", newline="") as f:
    f.write("Name,Age,Role\nSiva,25,Engineer\nPrasad,30,Manager\n")

with open("employees.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

### 8. Reading CSV Files as Dictionaries
Read CSV files using `csv.DictReader` to map columns directly to keys based on headers.

```python
import csv
with open("employees.csv", "r") as file:
    dict_reader = csv.DictReader(file)
    for row in dict_reader:
        print(f"Name: {row['Name']}, Role: {row['Role']}")
```

### 9. Writing CSV Files using DictWriter
Write structured dictionary lists to CSV files using `csv.DictWriter`.

```python
import csv
fieldnames = ["Product", "Price", "Stock"]
products = [
    {"Product": "Laptop", "Price": 999.99, "Stock": 10},
    {"Product": "Mouse", "Price": 29.99, "Stock": 50}
]

with open("inventory.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(products)

with open("inventory.csv", "r") as f:
    print(f.read())
```
