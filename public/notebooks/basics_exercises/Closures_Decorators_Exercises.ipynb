### Exercise 1: Custom Math Operator (Higher-Order Function)

**Task:** Write a function `apply_twice(func, val)` that:
- Takes a function `func` and a value `val` as arguments.
- Applies the function to the value twice (i.e., `func(func(val))`).
- Returns the result.

````python
# TODO: Implement apply_twice and test it with a helper function that adds 5
````

<details>
<summary>💡 Click to see solution</summary>

```python
def apply_twice(func, val):
    return func(func(val))

# Test

    return x + 5
```

### Exercise 2: Prefix Greeting Builder (Closure)

**Task:** Write a function `make_prefix_greeter(prefix)` that:
- Accepts a string `prefix` (e.g., "Hello", "Hi").
- Returns an inner function (closure) `greeter(name)`.
- The inner function should return a greeting string formatted as `"{prefix}, {name}!"`.

````python
# TODO: Implement make_prefix_greeter, create a 'Hi' greeter, and test it
````

<details>
<summary>💡 Click to see solution</summary>

```python
def make_prefix_greeter(prefix):
    def greeter(name):
        return f"{prefix}, {name}!"
    return greeter

# Test
greet_hi = make_prefix_greeter("Hi")
print(greet_hi("Aarav"))  # Output: Hi, Aarav!
print(greet_hi("Dev"))    # Output: Hi, Dev!
```

### Exercise 3: Simple Timer/Logger Decorator (Decorator)

**Task:** Write a decorator `@timer` that:
- Prints `"Start execution"` before calling the decorated function.
- Executes the decorated function.
- Prints `"End execution"` after calling the decorated function.

````python
# TODO: Implement timer decorator and apply to a test function
````

<details>
<summary>💡 Click to see solution</summary>

```python
def timer(func):
    def wrapper():
        print("Start execution")
        func()
        print("End execution")
    return wrapper

# Test
@timer
def task():
    print("Doing task...")

task()
```

### Exercise 4: Uppercase Argument Decorator

**Task:** Write a decorator `@uppercase_greet` that:
- Decorates a function that accepts a single string argument `name`.
- Converts the name argument to uppercase before passing it to the decorated function.

````python
# TODO: Implement uppercase_greet decorator and test with a greeting function
````

<details>
<summary>💡 Click to see solution</summary>

```python
def uppercase_greet(func):
    def wrapper(name):
        upper_name = name.upper()
        return func(upper_name)
    return wrapper

# Test
@uppercase_greet
def say_hello(name):
    print(f"Hello {name}")

say_hello("alice")  # Output: Hello ALICE
```
</details>

### Exercise 5: Call Counter (Closure + Decorator)

**Task:** Write a decorator `@count_calls` that:
- Uses a closure scope variable to count how many times the decorated function has been executed.
- Prints `"Call count: {count}"` on each execution.

````python
# TODO: Implement count_calls decorator and test it
````

<details>
<summary>💡 Click to see solution</summary>

```python
def count_calls(func):
    count = 0
    def wrapper(*args, **kwargs):
        nonlocal count
        count += 1
        print(f"Call count: {count}")
        return func(*args, **kwargs)
    return wrapper

# Test
@count_calls
def say_hello():
    print("Hello")

say_hello()
say_hello()
```
</details>
