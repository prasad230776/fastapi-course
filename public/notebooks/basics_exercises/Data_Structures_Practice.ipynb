# Python Data Structures - Practice
This notebook covers strings, lists, tuples, sets, dictionaries, and deques.

## Strings
### 1. Creating Strings
Strings can be created using single, double, or triple quotes.

```python
name = "Alice"
city = 'Hyderabad'
message = """Welcome
to Python"""
print(name)
print(city)
print(message)
```

### 2. String Indexing
Access individual characters by position. Python supports negative indexing starting from -1.

```python
text = "Python"
print(text[0])      # P
print(text[2])      # t
print(text[-1])     # n
```

### 3. String Slicing
Extract parts of a string using `start:stop:step`.

```python
text = "Python Programming"
print(text[:6])      # Python
print(text[7:])      # Programming
print(text[::2])     # Pto rgamn
print(text[::-1])    # Reverse
```

### 4. String Operators
Concatenate with `+`, repeat with `*`, and check membership with `in` / `not in`.

```python
print("Hello " + "Python")
print("Hi! " * 3)
print("Py" in "Python")
print("Java" not in "Python")
```

### 5. Common String Methods
Python provides a rich set of built-in methods for text manipulation.

```python
text = " hello python "
print(text.upper())
print(text.lower())
print(text.title())
print(text.strip())
print(text.replace("python", "World"))
print(text.find("python"))
print(text.count("o"))
```

### 6. Splitting and Joining Strings
Use `.split()` to break a string into a list, and `.join()` to combine list items back into a string.

```python
text = "Python,Java,C++"
languages = text.split(",")
print(languages)
print("-".join(languages))
```

### 7. String Formatting
Format strings dynamically using f-strings.

```python
name = "Alice"
age = 20
print(f"{name} is {age} years old.")
```

### 8. Strings are Immutable
Strings cannot be modified in-place; modifications create a new string object.

```python
text = "Python"
print(f"id={id(text)}  # id of text before modification")
text = "J" + text[1:]
print(f"id={id(text)}  # id of text after modification")
```

## Lists
### 9. Creating Lists
Lists are ordered, mutable, and can store elements of various data types.

```python
fruits = ["Apple", "Banana", "Orange"]
mixed = [1, "Python", 3.14, True]
empty = []
print(fruits)
print(mixed)
print(empty)
```

### 10. Accessing List Elements
Access elements via index, slice parts of the list, or reverse it.

```python
fruits = ["Apple", "Banana", "Orange", "Mango"]
print(fruits[0])      # Apple
print(fruits[-1])     # Mango
print(fruits[:2])     # ['Apple', 'Banana']
print(fruits[::-1])   # Reverse list
```

### 11. Modifying Lists (Adding and Extending)
Append single values, insert at specific indices, or extend with another list.

```python
numbers = [10, 20, 30]
numbers.append(40)
numbers.insert(1, 15)
numbers.extend([50, 60])
print(numbers)
```

### 12. Modifying Lists (Removing Elements)
Remove specific values, pop from the end (or index), or clear the list.

```python
numbers = [10, 15, 20, 30, 40, 50, 60]
numbers.remove(20)
popped_at_end = numbers.pop()
popped_at_pos_2 = numbers.pop(2)
print("Remaining:", numbers)
print("Popped item:", popped_at_end)
print("Popped item:", popped_at_pos_2)
```

### 13. Sorting and Reversing Lists
Sort the list in-place or reverse its elements.

```python
numbers = [10, 5, 20, 15]
numbers.sort()
print("Sorted:", numbers)
numbers.reverse()
print("Reversed:", numbers)
```

### 14. List Comprehensions
Write concise syntax to generate new lists.

```python
squares = [x * x for x in range(1, 6)]
print(squares)
```

## Tuples
### 15. Creating Tuples
Tuples are ordered and immutable. Remember the trailing comma for single-element tuples!

```python
numbers = (10, 20, 30)
mixed = (1, "Python", 3.14, True)
single = (10,)
empty = ()
print(numbers)
print(type(single))
```

### 16. Accessing Tuple Elements
Like lists, tuples support indexing and slicing.

```python
colors = ("Red", "Green", "Blue", "Yellow")
print(colors[0])      # Red
print(colors[-1])     # Yellow
print(colors[:2])     # ('Red', 'Green')
```

### 17. Tuple Packing and Unpacking
Unpack elements of a tuple into individual variables.

```python
student = ("Alice", 20, "Python")
name, age, course = student
print(name)
print(age)
print(course)
```

## Sets
### 18. Creating Sets
Sets are unordered, mutable collections of unique elements.

```python
fruits = {"Apple", "Banana", "Orange"}
numbers = {1, 2, 3, 4, 4, 3}  # duplicates removed
empty_set = set()              # empty set (NOT {})
print(fruits)
print(numbers)
```

### 19. Common Set Methods
Add or remove items from a set.

```python
numbers = {10, 20, 30}
numbers.add(40)
numbers.remove(20)
numbers.discard(50)  # no error if not present
print(numbers)
```

### 20. Set Mathematical Operations
Perform set union, intersection, difference, and symmetric difference.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Diff:", a ^ b)
```

## Dictionaries
### 21. Creating and Accessing Dictionaries
Store data as key-value pairs. Access using keys or the safer `.get()` method.

```python
student = {
    "name": "Alice",
    "age": 20,
    "course": "Python"
}
print(student["name"])
print(student.get("city", "Not Specified"))
```

### 22. Adding and Updating Dictionary Items
Create new key-value pairs or update existing ones.

```python
student = {"name": "Alice", "age": 20}
student["city"] = "Hyderabad"
student["age"] = 21
print(student)
```

### 23. Removing Dictionary Items
Remove specific keys using `.pop()` or the `del` keyword.

```python
student = {"name": "Alice", "age": 21, "city": "Hyderabad", "course": "Python"}
student.pop("city")
del student["course"]
print(student)
```

### 24. Dictionary Keys, Values, and Items
Access list-like views of the keys, values, or key-value pairs.

```python
student = {"name": "Alice", "age": 21}
print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

### 25. Iterating Through Dictionaries
Loop through keys and values.

```python
student = {"name": "Alice", "age": 21}
for key, value in student.items():
    print(key, ":", value)
```

### 26. Dictionary Comprehensions
Construct dictionaries dynamically and concisely.

```python
squares = {x: x * x for x in range(1, 6)}
print(squares)
```

## Deque
### 27. Creating a Deque
Deques are optimized double-ended queues supporting O(1) appends and pops from both ends.

```python
from collections import deque
dq = deque([10, 20, 30])
print(dq)
```

### 28. Deque Operations
Perform append, appendleft, pop, and popleft.

```python
from collections import deque
dq = deque([10, 20, 30])
dq.append(40)
dq.appendleft(5)
print("After appends:", dq)
dq.pop()
dq.popleft()
print("After pops:", dq)
```

### 29. Rotating a Deque
Shift deque elements circularly to the right (positive steps) or left (negative steps).

```python
from collections import deque
dq = deque([1, 2, 3, 4])
dq.rotate(1)
print("Rotated 1:", dq)
dq.rotate(-2)
print("Rotated -2:", dq)
```
