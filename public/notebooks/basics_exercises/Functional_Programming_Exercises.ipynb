#### Exercise 1: Custom Lambda Sorting

**Task:** Given a list of dictionaries representing products, sort the list by price in descending order using a lambda function.

````python
products = [
    {"name": "Laptop", "price": 1200},
    {"name": "Mouse", "price": 25},
    {"name": "Keyboard", "price": 80},
    {"name": "Monitor", "price": 300}
]
# TODO: Sort the products list by price in descending order and print the result
````

<details>
<summary>💡 Click to see solution</summary>

```python
products = [
    {"name": "Laptop", "price": 1200},
    {"name": "Mouse", "price": 25},
    {"name": "Keyboard", "price": 80},
    {"name": "Monitor", "price": 300}
]

sorted_products = sorted(products, key=lambda item: item["price"], reverse=True)
print(sorted_products)
# Output: [{'name': 'Laptop', 'price': 1200}, {'name': 'Monitor', 'price': 300}, {'name': 'Keyboard', 'price': 80}, {'name': 'Mouse', 'price': 25}]
```
</details>

#### Exercise 2: String Length Mapping

**Task:** Use the `map()` function to convert a list of words into a list containing the lengths of each word.

````python
words = ["functional", "programming", "in", "python", "is", "fun"]
# TODO: Use map to get lengths of each word and print as a list
````

<details>
<summary>💡 Click to see solution</summary>

```python
words = ["functional", "programming", "in", "python", "is", "fun"]
word_lengths = list(map(lambda word: len(word), words))

print(word_lengths)  # Output: [10, 11, 2, 6, 2, 3]
```
</details>

#### Exercise 3: Vowel Filtering

**Task:** Use the `filter()` function to extract only the words from a list that start with a vowel (A, E, I, O, U, case-insensitive).

````python
words_list = ["Apple", "Banana", "Orange", "kiwi", "Umbrella", "grape", "Igloo"]
# TODO: Use filter to extract words starting with a vowel and print as a list
````

<details>
<summary>💡 Click to see solution</summary>

```python
words_list = ["Apple", "Banana", "Orange", "kiwi", "Umbrella", "grape", "Igloo"]
vowel_words = list(filter(lambda word: word[0].lower() in 'aeiou', words_list))

print(vowel_words)  # Output: ['Apple', 'Orange', 'Umbrella', 'Igloo']
```
</details>

#### Exercise 4: Calculating Product using Reduce

**Task:** Use `reduce()` from the `functools` module to compute the product of all elements in a list of numbers.

````python
from functools import reduce
numbers = [1, 2, 3, 4, 5, 6]
# TODO: Calculate the product of the numbers list using reduce
````

<details>
<summary>💡 Click to see solution</summary>

```python
from functools import reduce
numbers = [1, 2, 3, 4, 5, 6]
product = reduce(lambda x, y: x * y, numbers)

print(product)  # Output: 720
```
</details>

#### Exercise 5: Group Assessment Validation

**Task:** Given a list of student scores, use `any()` and `all()` to verify:
1. If anyone has scored a perfect 100.
2. If all students have passed (passed score >= 50).

````python
scores1 = [55, 78, 92, 100, 45]
scores2 = [60, 75, 80, 52, 90]
# TODO: Print True/False for the two checks on both scores lists
````

<details>
<summary>💡 Click to see solution</summary>

```python
scores1 = [55, 78, 92, 100, 45]
scores2 = [60, 75, 80, 52, 90]

# 1. Has anyone scored 100?
any_100_s1 = any(score == 100 for score in scores1)
any_100_s2 = any(score == 100 for score in scores2)
print("Scores1 has 100:", any_100_s1)  # Output: True
print("Scores2 has 100:", any_100_s2)  # Output: False

# 2. Did all students pass (score >= 50)?
all_pass_s1 = all(score >= 50 for score in scores1)
all_pass_s2 = all(score >= 50 for score in scores2)
print("Scores1 all pass:", all_pass_s1)  # Output: False
print("Scores2 all pass:", all_pass_s2)  # Output: True
```
</details>
