### 1. Everything is an Object & Variables as References

In Python, everything is treated as an object (integers, floats, strings, functions, modules). Variables do not store values directly; they store **references** (pointers) to the location in memory where the object is stored.

#### Object Identity & Reference Sharing Example

````python
x = [1, 2, 3]
y = x

print(f"ID of x: {id(x)}")
print(f"ID of y: {id(y)}")
print("x is y:", x is y)
````

#### Value Equality vs Object Identity Example

````python
x = [1, 2, 3]
z = [1, 2, 3]

print("x == z (value comparison):", x == z)
print("x is z (identity comparison):", x is z)
````

### 2. Integer Value Caching (Integer Interning)

CPython pre-allocates and caches all integers in the range `[-5, 256]` at startup. Referencing any integer in this range returns a reference to the existing cached integer object.

#### Small Integer Cache Example

````python
a = 100
b = 100

print("a is b:", a is b)
print("Address match:", id(a) == id(b))
````

#### Outside Cache Range Example

````python
x = 300
y = 300

print("x is y:", x is y)
````

### 3. Mutability (Mutable vs Immutable Objects)

In Python, every object is either mutable (can be changed in-place) or immutable (cannot be changed in-place).

#### Immutable Objects Example

````python
x = 10
print(f"Initial ID of x: {id(x)}")

x = x + 5
print(f"New ID of x: {id(x)}")
````

#### Mutable Objects Example

````python
lst = [1, 2, 3]
print(f"Initial ID of lst: {id(lst)}")

lst.append(4)
print(f"New ID of lst: {id(lst)}")
print(f"List content: {lst}")
````

#### Mutable Objects Inside Immutable Containers Nuance

````python
t = (1, 2, [3, 4])
t[2].append(5)
print(t)
````

### 4. Floating Point Precision Nuances

Floating-point numbers in computers are represented as binary fractions. This leads to some surprising behavior when performing decimal calculations because repeating fractions cannot be stored exactly.

#### The Classic Floating-Point Issue: `0.1 + 0.2`

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

#### Why `0.5 + 0.25 == 0.75` Works

````python
print(0.5 + 0.25)
print(0.5 + 0.25 == 0.75)
````

#### Handling Exact Decimal Math (Decimal Module)

````python
from decimal import Decimal

val1 = Decimal('0.1')
val2 = Decimal('0.2')
print(val1 + val2)
print(val1 + val2 == Decimal('0.3'))
````

### 8. Loop-Else Clauses (`for-else` and `while-else`)

Unlike many programming languages, Python allows an optional `else` block after loops. The `else` block executes **only when the loop completes normally without encountering a `break` statement**.

#### Example 1: Searching for a Divisor

````python
number = 21

for i in range(2, number):
    if number % i == 0:
        print(f"{i} is a divisor.")
        break
else:
    print("No divisor found.")
````

#### Example 2: Checking Whether a Number is Prime

````python
number = 29

for i in range(2, int(number ** 0.5) + 1):
    if number % i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
````

#### `while-else`

````python
count = 1

while count <= 5:
    print(count)
    count += 1
else:
    print("Loop Finished")
````

#### Keyword-only Parameters (`*`)

Parameters appearing after `*` must always be passed using keyword arguments.

````python
def register(name, *, city, email):
    print(name)
    print(city)
    print(email)

register(
    "Alice",
    city="Hyderabad",
    email="alice@test.com"
)
````

````python
def connect(host, *, port, timeout):
    print(host)
    print(port)
    print(timeout)

connect(
    "localhost",
    port=8000,
    timeout=30
)
````

#### Combining Both

A function can use positional-only and keyword-only parameters together.

````python
def calculate(a, b, /, *, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b

print(calculate(10, 20, operation="add"))
````

### 10. Packing and Unpacking

The operators `*` and `**` have two complementary roles in Python:
- **Packing** collects multiple values into a single variable.
- **Unpacking** expands a collection into individual values.

#### Packing Positional Arguments (`*args`)

`*args` collects all remaining positional arguments into a tuple.

````python
def display(*values):
    print(values)

display("Python")
display("Python", "Java", "C++")
````

````python
def total(*numbers):
    result = 0
    for number in numbers:
        result += number
    return result

print(total(10, 20))
print(total(10, 20, 30))
print(total(10, 20, 30, 40))
````

#### Packing Keyword Arguments (`**kwargs`)

`**kwargs` collects keyword arguments into a dictionary.

````python
def display(**details):
    print(details)

display(name="Alice", city="Hyderabad")
````

````python
def display(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

display(
    name="Alice",
    city="Hyderabad",
    course="Python"
)
````
