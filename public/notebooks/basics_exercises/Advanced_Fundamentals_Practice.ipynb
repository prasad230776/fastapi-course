### 1. Advanced String Indexing and Slicing

Strings are sequences of characters. Python allows extracting portions of a string using indexing and slicing:

```python
string[start:stop:step]
```

- **start** – Starting index (inclusive)
- **stop** – Ending index (exclusive)
- **step** – Number of characters to skip

Any of these values may be omitted.

#### Basic Slicing

````python
text = "Advanced Python"

print(text[0])
print(text[3])
print(text[-1])
print(text[-2])
````

#### Extracting a Portion of a String

````python
text = "Advanced Python"

print(text[0:8])
print(text[9:15])
````

#### Omitting Start or Stop

````python
text = "Advanced Python"

print(text[:8])
print(text[9:])
````

#### Using Step

````python
text = "Programming"

print(text[::2])
print(text[1::2])
````

#### Reversing a String

A negative step traverses the string from right to left.

````python
text = "Python"

print(text[::-1])
````

#### Practical Example

Extract the file extension.

````python
filename = "report.pdf"

print(filename[-3:])
````

### 2. Python Nuances

Python provides several elegant language features that simplify common programming tasks.

#### Short-Circuit Evaluation

Logical operators don't always evaluate every expression.

##### Using `or` (Default/Fallback Values)

````python
username = ""

if username:
    print(username)
else:
    print("Guest")
````

````python
username = ""
print(username or "Guest")
````

````python
username = "Alice"
print(username or "Guest")
````

##### Using `and` (Conditional Execution)

````python
logged_in = True

if logged_in:
    print("Welcome!")
````

````python
logged_in = True
logged_in and print("Welcome!")
````

### 3. Loop-Else Clauses (`for-else` and `while-else`)

Unlike many programming languages, Python allows an optional `else` block after loops. The `else` block executes **only when the loop completes normally without encountering a `break` statement**.

#### Example 1: Searching for a Divisor

````python
number = 21

for i in range(2, number):
    if number % i == 0:
        print(f"{i} is a divisor.")
        break
else:
    print("No divisor found.")
````

#### Example 2: Checking Whether a Number is Prime

````python
number = 29

for i in range(2, int(number ** 0.5) + 1):
    if number % i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
````

#### `while-else`

````python
count = 1

while count <= 5:
    print(count)
    count += 1
else:
    print("Loop Finished")
````

### 4. Function Arguments

Functions become more flexible when they can accept arguments in different ways. Python supports positional arguments, keyword arguments, positional-only parameters, and keyword-only parameters.

````python
def introduce(name, city):
    print(f"My name is {name}.")
    print(f"I live in {city}.")

introduce("Alice", "Hyderabad")
````

````python
# Order matters for positional arguments
introduce("Hyderabad", "Alice")
````

````python
def power(base, exponent, /):
    return base ** exponent

print(power(2, 5))
````

#### Keyword-only Parameters (`*`)

Parameters appearing after `*` must always be passed using keyword arguments.

````python
def register(name, *, city, email):
    print(name)
    print(city)
    print(email)

register(
    "Alice",
    city="Hyderabad",
    email="alice@test.com"
)
````

````python
def connect(host, *, port, timeout):
    print(host)
    print(port)
    print(timeout)

connect(
    "localhost",
    port=8000,
    timeout=30
)
````

#### Combining Both

A function can use positional-only and keyword-only parameters together.

````python
def calculate(a, b, /, *, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b

print(calculate(10, 20, operation="add"))
````

### 5. Packing and Unpacking

The operators `*` and `**` have two complementary roles in Python:
- **Packing** collects multiple values into a single variable.
- **Unpacking** expands a collection into individual values.

#### Packing Positional Arguments (`*args`)

`*args` collects all remaining positional arguments into a tuple.

````python
def display(*values):
    print(values)

display("Python")
display("Python", "Java", "C++")
````

````python
def total(*numbers):
    result = 0
    for number in numbers:
        result += number
    return result

print(total(10, 20))
print(total(10, 20, 30))
print(total(10, 20, 30, 40))
````

#### Packing Keyword Arguments (`**kwargs`)

`**kwargs` collects keyword arguments into a dictionary.

````python
def display(**details):
    print(details)

display(name="Alice", city="Hyderabad")
````

````python
def display(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

display(
    name="Alice",
    city="Hyderabad",
    course="Python"
)
````
