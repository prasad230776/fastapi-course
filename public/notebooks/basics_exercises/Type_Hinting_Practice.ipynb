# Type Hinting - Practice
This notebook covers variable annotations, function signatures, collection types, and modern Union/Optional syntax.

### 1. Basic Variable Annotations
Annotate variables using a colon followed by their type name.

```python
age: int = 25
name: str = "Alice"
is_active: bool = True
price: float = 99.99
print(name, age, is_active, price)
```

### 2. Function Parameter and Return Annotations
Annotate parameters and define function return values using the `->` symbol.

```python
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

print("Total:", calculate_total(99.99, 3))
```

### 3. Collection Type Hints (Lists, Dicts, Sets, Tuples)
Use modern Python 3.9+ built-in collections directly to describe element types.

```python
prices: list[float] = [10.99, 5.50, 20.00]
user_ages: dict[str, int] = {"Alice": 25, "Bob": 30}
unique_names: set[str] = {"Alice", "Bob"}
coordinates: tuple[float, float] = (12.34, 56.78)

print(prices)
print(user_ages)
print(unique_names)
print(coordinates)
```

### 4. Modern Union Type Syntax (|)
Starting with Python 3.10, represent multiple possible types using the `|` operator.

```python
result: int | float = 10.5
print(result)
```

### 5. Modern Optional Type Syntax (str | None)
Represent nullable or optional values using the `| None` syntax.

```python
middle_name: str | None = None
print(middle_name)
```
