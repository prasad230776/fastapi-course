### 1. Function Arguments: Positional and Keyword passing

Arguments can be passed to functions either based on position (order) or keyword (parameter name).

````python
def display_profile(name, age):
    print(f"Name: {name}, Age: {age}")

# Positional
display_profile("Alice", 25)

# Keyword
display_profile(age=25, name="Alice")
````

### 3. Packing and Unpacking

Using `*args` and `**kwargs` to capture variable numbers of arguments, and using `*` / `**` to unpack lists/dicts.

#### Packing Example (`*args` and `**kwargs`)

````python
def func(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

func(1, 2, name="Alice", age=25)
````

#### Unpacking Example (`*` and `**`)

````python
def add(x, y, z):
    return x + y + z

nums = [1, 2, 3]
print("Unpacked List:", add(*nums))

details = {"x": 10, "y": 20, "z": 30}
print("Unpacked Dict:", add(**details))
````

### 4. Mutable Default Argument Pitfall

Default arguments in Python are evaluated only once when the function is defined. If you use a mutable object as a default argument, it persists across subsequent calls.

#### The Pitfall in Action

````python
def add_item(item, basket=[]):
    basket.append(item)
    return basket

print(add_item("Apple"))
print(add_item("Banana"))  # Notice that Apple is still there!
````

#### The Solution using `None`

````python
def add_item_clean(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(add_item_clean("Apple"))
print(add_item_clean("Banana"))  # New list instantiated properly!
````

### 5. Variable Scope

Variable resolution in Python uses the LEGB rule (Local, Enclosing, Global, Built-in).

#### Local Scope

````python
def calculate():
    result = 100
    print("Inside:", result)

calculate()
````

#### Global Scope and the `global` Keyword

````python
count = 0

def increment():
    global count
    count += 1

increment()
print("Global count:", count)
````

#### Enclosing Scope and `nonlocal` Keyword

````python
def outer():
    count = 10

    def inner():
        nonlocal count
        count += 5
        print("Inner count:", count)

    inner()
    print("Outer count:", count)

outer()
````
