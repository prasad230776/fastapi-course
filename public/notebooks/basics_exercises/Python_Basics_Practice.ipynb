# Python Basics - Follow-Along Practice
This notebook covers variables, data types, operators, and control flow in Python.

### 1. Variables and Multiple Assignment
Variables store references to objects. In Python, you can define variables and assign multiple values in a single line.

```python
# Variables and multiple assignment
name = "Alice"
age = 20
price = 99.99
is_active = True

x, y, z = 10, 20, 30
print(name, age, price, is_active)
print(x, y, z)
```

### 2. Variable Swapping
Python allows a clean syntax to swap two variables without needing a temporary third variable.

```python
a, b = 10, 20
a, b = b, a
print("a:", a, "b:", b)
```

### 3. Object Identity and Reference
Use `==` to compare values and `is` to check if two variables point to the exact same object in memory.

```python
a = [1, 2]
b = a
c = [1, 2]

print("a == b:", a == b)   # True (same values)
print("a is b:", a is b)   # True (same object reference)
print("a == c:", a == c)   # True (same values)
print("a is c:", a is c)   # False (different object in memory)
```

### 4. Small Integer Caching
Python caches small integers from -5 to 256 for optimization. Compare with larger integers.

```python
a = 100
b = 100
print("100 is 100:", a is b)      # True

x = 1000
y = 1000
print("1000 is 1000:", x is y)    # False (in standard interpreter, references may vary)
```

### 5. Floating-Point Precision
Floating-point numbers can represent binary fractions only approximately, causing minor precision differences.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)  # False
import math
print("Is close:", math.isclose(0.1 + 0.2, 0.3)) # True
```

### 6. Data Types and Type Conversion
Python is dynamically typed. Check types using `type()` and convert types using casting functions.

```python
print(type(10))
print(type(3.14))
print(type("Hello"))

# Conversion
print(int("10"))
print(float("3.14"))
print(str(100))
print(list("Python"))
```

### 7. Arithmetic Operators
Perform basic mathematical operations.

```python
a = 10
b = 3
print("a + b =", a + b)   # Addition
print("a - b =", a - b)   # Subtraction
print("a * b =", a * b)   # Multiplication
print("a / b =", a / b)   # Division
print("a // b =", a // b) # Floor Division
print("a % b =", a % b)   # Modulus (Remainder)
print("a ** b =", a ** b) # Exponent
```

### 8. Comparison and Logical Operators
Use comparison operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) and logical operators (`and`, `or`, `not`) to evaluate conditions.

```python
a = 10
b = 3
print("a > b:", a > b)
print("a == b:", a == b)

age = 20
has_id = True
print("Can enter:", age >= 18 and has_id)
print("Cannot enter:", not has_id)
```

### 9. Truthy and Falsy Values
Empty sequences, zero values, `None`, and `False` are Falsy. Almost all other objects are Truthy.

```python
print("bool(0):", bool(0))
print("bool(''):", bool(""))
print("bool([]):", bool([]))
print("bool('Python'):", bool("Python"))

# Short-circuit evaluation
print("3 and 4:", 3 and 4)
print("0 or 'Hi':", 0 or "Hi")
name = "" or "Guest"
print("Name:", name)
```

### 10. Assignment Operators
Update variables directly using shorthand assignment operators.

```python
x = 10
x += 5
print("x += 5:", x)
x *= 2
print("x *= 2:", x)
```

### 11. Identity and Membership Operators
Check object identity with `is`/`is not` and collection membership with `in`/`not in`.

```python
a = 20
b = 20
print("a is b:", a is b)
print("'a' in 'apple':", "a" in "apple")
print("3 in [1, 2, 3]:", 3 in [1, 2, 3])
```

### 12. Bitwise Operators
Operate directly on the binary representation of integers.

```python
a = 5  # 0101
b = 3  # 0011
print("a & b:", a & b)    # AND
print("a | b:", a | b)    # OR
print("a ^ b:", a ^ b)    # XOR
print("a << 1:", a << 1)  # Left Shift (multiply by 2)
print("a >> 1:", a >> 1)  # Right Shift (divide by 2)
```

### 13. Conditional Statements (if, elif, else, match)
Control execution flow based on conditions.

```python
marks = 82
if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
else:
    print("Grade: C")
```

```python
# match...case (Python 3.10+)
day = 2
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case _:
        print("Other Day")
```

### 14. Loops and Control Statements
Use `for` and `while` loops to repeat actions, and control them using `break`, `continue`, and `pass`.

```python
# for loop
for i in range(3):
    print("for:", i)
```

```python
# while loop
count = 1
while count <= 3:
    print("while:", count)
    count += 1
```

```python
# break, continue, pass
for i in range(5):
    if i == 2:
        continue
    if i == 4:
        break
    print("loop control:", i)
```

### 15. Loop Else Clauses
The `else` block in a loop executes if the loop completes normally without encountering a `break`.

```python
for i in range(3):
    print("item:", i)
else:
    print("Loop completed successfully!")
```

```python
for i in range(3):
    if i == 1:
        print("Break out!")
        break
else:
    print("This won't be printed.")
```
