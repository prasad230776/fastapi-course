### 1. Exception Handling

An exception is a runtime error that would normally crash a program. We can intercept exceptions using a `try-except` block.

#### The `try-except-else-finally` Block Example

````python
try:
    num = int(input("Enter a number: ")) # Try entering 0 or a string like "two"
    result = 100 / num
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except ValueError:
    print("Error: Please enter a valid integer!")
else:
    print(f"Success! Result is {result}")
finally:
    print("Execution of block completed.")
````

#### Raising Exceptions (`raise`) Example

````python
def check_positive(value):
    if value < 0:
        raise ValueError("Negative values are not allowed!")
    return value

try:
    check_positive(-10)
except ValueError as e:
    print("ValueError caught:", e)
````

### 2. Custom Exceptions

We can define our own custom exception types by creating classes that inherit from Python's base `Exception` class.

#### Custom Exception Inheritance Example

````python
class InsufficientFundsError(Exception):
    """Exception raised when an account balance is too low for a transaction."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Attempted to withdraw ${amount} but balance is only ${balance}.")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount

try:
    withdraw(50, 100)
except InsufficientFundsError as e:
    print(f"Transaction Declined: {e}")
````

### 3. Data Handling (File I/O)

We can use Python's built-in tools to read and write text files, CSV tables, and JSON database payloads.

#### Text Files with Context Managers

````python
# Writing text
with open("note.txt", "w") as file:
    file.write("Hello World!\nWelcome to data handling.")

# Reading text
with open("note.txt", "r") as file:
    content = file.read()
    print(content)
````

#### CSV Files (Writing & Reading)

````python
import csv

# Writing rows
data = [
    ["Name", "Age", "Role"],
    ["Alice", 25, "Developer"],
    ["Bob", 30, "Manager"]
]
with open("users.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

# Reading rows
with open("users.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)

# Reading rows as dictionaries
with open("users.csv", "r") as file:
    dict_reader = csv.DictReader(file)
    for row in dict_reader:
        print(f"{row['Name']} works as a {row['Role']}")
````

#### JSON Data (Serialization and Deserialization)

Python dictionaries can be serialized into JSON string payloads, or deserialized back into Python types.

````python
import json

student = {
    "name": "Alice",
    "age": 25,
    "courses": ["Python", "FastAPI"]
}

# Serialize (dict -> JSON string)
json_string = json.dumps(student, indent=2)
print("JSON String:\n", json_string)

# Deserialize (JSON string -> dict)
parsed_dict = json.loads(json_string)
print("Parsed age:", parsed_dict["age"])

# Write JSON to file
with open("data.json", "w") as file:
    json.dump(student, file, indent=4)

# Read JSON from file
with open("data.json", "r") as file:
    loaded_data = json.load(file)
    print("Loaded Name:", loaded_data["name"])
````
