### 1. Simple Class and Instance Blueprint
Define a class blueprint with `__init__` constructor, attributes, and instance methods.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def study(self):
        print(f"{self.name} is studying.")

    def introduce(self):
        print(f"My name is {self.name}. I am {self.age} years old.")

student1 = Student("Alice", 20)
student1.introduce()
student1.study()
```

### 2. Instance Methods, Class Methods, and Static Methods
Compare instance methods (operating on `self`), class methods (operating on `cls`), and static methods (utility functions).

```python
class Account:
    bank_name = "State Bank of India"

    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount: float):
        self.balance += amount
        return self.balance

    @classmethod
    def update_bank(cls, new_name: str):
        cls.bank_name = new_name

    @staticmethod
    def is_valid_amount(amount: float) -> bool:
        return amount > 0

acc = Account("Alice", 1000.0)
acc.deposit(500.0)
Account.update_bank("SBI Bank")
print("Balance:", acc.balance)
print("Bank Name:", Account.bank_name)
print("Is Valid:", Account.is_valid_amount(-10))
```

### 3. Type Checking: type vs isinstance
Check object types exactly using `type()`, or check hierarchy/subclasses using `isinstance()`.

```python
class Animal: pass
class Dog(Animal): pass

buddy = Dog()
print("type(buddy) is Dog:", type(buddy) is Dog)
print("type(buddy) is Animal:", type(buddy) is Animal)
print("isinstance(buddy, Animal):", isinstance(buddy, Animal))
```

### 4. Basic Inheritance
Inherit attributes and methods from a parent class, and call `super().__init__()` to initialize them.

```python
class Animal:
    def __init__(self, name: str):
        self.name = name
    def eat(self):
        return f"{self.name} is eating"

class Dog(Animal):
    def __init__(self, name: str, breed: str):
        super().__init__(name)
        self.breed = breed
    def bark(self):
        return "Woof!"

buddy = Dog("Buddy", "Golden Retriever")
print(buddy.eat())
print(buddy.bark())
```

### 5. Custom Subclass Hooks (__init_subclass__)
Pass arguments directly along with the parent class name in class declarations to configure subclass attributes dynamically.

```python
class DatabaseModel:
    table_name: str
    def __init_subclass__(cls, table: str, **kwargs):
        super().__init_subclass__(**kwargs)
        cls.table_name = table

class UserProfile(DatabaseModel, table="users"):
    pass

print("Table Name:", UserProfile.table_name)
```

### 6. Encapsulation: Private & Protected Fields
Enforce internal/subclass visibility using single underscores (protected convention) or double underscores (private/name-mangled).

```python
class User:
    def __init__(self, username: str, pin: str):
        self.username = username
        self._status = "active"
        self.__pin = pin

u = User("alice", "1234")
print("username:", u.username)
print("protected status:", u._status)
try:
    print(u.__pin)
except AttributeError as e:
    print("Private access failed:", e)
print("Mangled pin:", u._User__pin)
```

### 7. Properties: Getters & Setters
Control access and validate values of a field using `@property` getter and `.setter` decorators.

```python
class Product:
    def __init__(self, name: str, price: float):
        self.name = name
        self._price = price

    @property
    def price(self) -> float:
        return self._price

    @price.setter
    def price(self, value: float):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

item = Product("Laptop", 999.0)
item.price = 1050.0
print("Price:", item.price)
try:
    item.price = -50.0
except ValueError as e:
    print("Validation failed:", e)
```

### 8. Abstract Base Classes (ABCs)
Define common interfaces that subclasses must implement using the `abc` module.

```python
from abc import ABC, abstractmethod

class BaseRepository(ABC):
    @abstractmethod
    def save(self, data: dict) -> None:
        pass

class SQLRepository(BaseRepository):
    def save(self, data: dict) -> None:
        print(f"Saving {data} to SQL")

repo = SQLRepository()
repo.save({"id": 1})
```

### 9. Introspection using __dict__ and __annotations__
Inspect class types, variable mappings, and annotations dynamically at runtime.

```python
class User:
    username: str
    email: str
    def __init__(self, username: str, email: str):
        self.username = username
        self.email = email

print("Annotations:", User.__annotations__)
u = User("alice", "alice@example.com")
print("Instance namespace dict:", u.__dict__)
```
