### 1. Python Modules

A module is simply a `.py` file containing functions, classes, or variables. You can import modules in three ways.

#### Creating a Module File
Let's write a file named `utils.py` using Jupyter's `%%writefile` magic command.

````python
%%writefile utils.py
def greet(name):
    return f"Hello, {name}!"

PI = 3.14159
````

#### A. Standard Import

````python
import utils

print(utils.greet("Alice"))
print(utils.PI)
````

#### B. Importing Specific Elements

````python
from utils import greet, PI

print(greet("Bob"))
print(PI)
````

#### C. Alias Import (Renaming)

````python
import utils as u
from utils import greet as hello

print(u.PI)
print(hello("Charlie"))
````

### 2. Python Packages

Packages allow structuring multiple modules into folders. A directory is turned into a package by adding a `__init__.py` file.

#### Creating a Package Structure
Let's create a package folder called `db` containing queries.

````python
import os
from pathlib import Path

# Create directory db
Path("db").mkdir(exist_ok=True)

# Create empty __init__.py
Path("db/__init__.py").touch()

# Create connection.py
with open("db/connection.py", "w") as f:
    f.write('def get_db_connection():\n    return "Database Connected!"\n')

# Create queries.py
with open("db/queries.py", "w") as f:
    f.write('def fetch_users():\n    return ["Alice", "Bob", "Charlie"]\n')
````

#### Importing from a Package

````python
from db.connection import get_db_connection
from db.queries import fetch_users

print(get_db_connection())
print(fetch_users())
````

### 3. Prominent Standard Libraries

Python has built-in modules to interact with numbers, OS environments, files, and randomness.

#### The `math` Module

````python
import math

print("PI constant:", math.pi)
print("Square root of 16:", math.sqrt(16))
print("Ceiling of 4.2:", math.ceil(4.2))
print("Floor of 4.8:", math.floor(4.8))
````

#### The `random` Module

````python
import random

print("Random float [0, 1):", random.random())
print("Random int [1, 10]:", random.randint(1, 10))
print("Random choice from list:", random.choice(["red", "blue", "green"]))

numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("Shuffled list:", numbers)
````

#### The `os` Module

````python
import os

print("Current working directory:", os.getcwd())
print("PATH environment variable length:", len(os.environ.get("PATH", "")))
````

#### The `pathlib` Module (Recommended for Paths)

````python
from pathlib import Path

current_path = Path.cwd()
print("Current folder path:", current_path)

logs_dir = current_path / "logs"
if not logs_dir.exists():
    logs_dir.mkdir()
    print("Logs folder created!")

report_file = Path("report.csv")
print("Suffix:", report_file.suffix)
print("Stem:", report_file.stem)
````

### 4. Configuration with `python-dotenv`

In production, configuration and secrets must be loaded from environment variables rather than hardcoded in the codebase.

#### Creating a `.env` File

````python
%%writefile .env
DATABASE_URL=postgresql://user:password@localhost:5432/mydb
PORT=8000
DEBUG=True
API_KEY=secret_token_123
````

#### Loading Variables

````python
import os
from dotenv import load_dotenv

# Load .env file
load_dotenv()

db_url = os.environ.get("DATABASE_URL")
port = int(os.environ.get("PORT", 8000))

print("Loaded Database URL:", db_url)
print("Loaded Port:", port)
````
