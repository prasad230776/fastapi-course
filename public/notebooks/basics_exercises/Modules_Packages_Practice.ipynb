### 1. Creating and Importing a Module
Import functionality from one file (module) into another. Here we write a small script file to act as a module, then import it.

  *Please practice using .py file in some editor, create a calculator_module.py file 

```python
# Write a mock module file
with open('calculator_module.py', 'w') as f:
    f.write('def add(a, b):\n    return a + b\n\nPI = 3.14159\n')

# Import and use the module
import calculator_module
print("add(5, 3):", calculator_module.add(5, 3))
print("PI:", calculator_module.PI)
```

### 2. Using pathlib for Paths
Manage paths in an object-oriented way using the standard library's `pathlib.Path`.

```python
from pathlib import Path
current_dir = Path.cwd()
print("Current directory:", current_dir)
file_path = current_dir / "test_pathlib.txt"
file_path.write_text("Hello Pathlib!")
print("File contents:", file_path.read_text())
file_path.unlink() # Cleanup
```

### 3. Using os Module
Interact with the operating system, fetch workspace directories, and get environment variables.

```python
import os
print("OS cwd:", os.getcwd())
print("Database URL:", os.environ.get("DATABASE_URL", "sqlite:///local.db"))
```

### 4. Using datetime Module
Generate, format, and parse dates and times.

```python
import datetime
now = datetime.datetime.now()
print("Now:", now)
formatted = now.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted:", formatted)
parsed = datetime.datetime.strptime("2026-07-20", "%Y-%m-%d")
print("Parsed:", parsed)
```

### 5. Using uuid Module
Generate RFC 4122 compliant universally unique identifiers.

```python
import uuid
unique_id = uuid.uuid4()
print("Generated UUID:", unique_id)
```

### 6. Using json Module
Serialize Python dictionaries to JSON strings and deserialize JSON strings back into dictionaries.

```python
import json
data = {"name": "Alice", "role": "admin"}
json_str = json.dumps(data)
print("Serialized:", json_str)
parsed_data = json.loads(json_str)
print("Deserialized:", parsed_data)
```

### 7. Making GET requests with requests
Perform HTTP GET requests to fetch data from APIs.

```python
!pip install requests
```

```python
import requests
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")
print("GET Status Code:", response.status_code)
if response.status_code == 200:
    print("Title:", response.json()["title"])
```

### 8. Making POST requests with requests
Perform HTTP POST requests to submit data to APIs.

```python
import requests
payload = {
    "title": "Hello FastAPI",
    "body": "Learning about modules and packages",
    "userId": 1
}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=payload)
print("POST Status Code:", response.status_code)
print("Response:", response.json())
```
