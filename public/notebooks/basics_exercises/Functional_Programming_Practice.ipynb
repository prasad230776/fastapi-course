## 1. Procedural vs Functional Approach

In functional programming, we describe what transformation should happen rather than how to perform it step by step.

#### Comparison Example

```python
numbers = [1, 2, 3, 4, 5]

# Procedural Approach
procedural_result = []
for number in numbers:
    if number % 2 == 0:
        procedural_result.append(number * number)
print("Procedural:", procedural_result)

# Functional Approach
functional_result = map(
    lambda x: x * x,
    filter(lambda x: x % 2 == 0, numbers)
)
print("Functional:", list(functional_result))
```

## 2. Lambda Functions

A lambda function is a small anonymous function created using the `lambda` keyword.

#### Lambda Function Examples

```python
# Single parameter
square = lambda x: x * x
print("Square of 5:", square(5))

# Multiple parameters
add = lambda a, b: a + b
print("Sum of 10 and 20:", add(10, 20))
```

#### Lambda with Custom Sorting

```python
students = [
    ("John", 82),
    ("Alice", 95),
    ("Bob", 76)
]

# Sort by marks (second element of tuple)
students.sort(key=lambda student: student[1])
print("Sorted students:", students)
```

## 3. Built-in Higher-Order Functions

Python provides functions like `map()`, `filter()`, and `reduce()` to simplify common data transformations.

#### The `map()` Function

```python
numbers = [1, 2, 3, 4, 5]

# Map with lambda
squares = map(lambda x: x * x, numbers)
print("Squares list:", list(squares))

# Map multiple iterables
numbers1 = [1, 2, 3]
numbers2 = [10, 20, 30]
sums = map(lambda x, y: x + y, numbers1, numbers2)
print("Sums list:", list(sums))
```

#### The `filter()` Function

```python
numbers = [1, 2, 3, 4, 5, 6]

# Filter even numbers
evens = filter(lambda number: number % 2 == 0, numbers)
print("Even numbers:", list(evens))
```

#### The `reduce()` Function

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Calculate sum
total = reduce(lambda a, b: a + b, numbers)
print("Total sum:", total)

# Calculate product
product = reduce(lambda a, b: a * b, [2, 3, 4])
print("Product:", product)
```

## 4. Other Higher-Order Functions

Functions like `sorted()`, `any()`, and `all()` help with collections and conditions.

#### Custom Sorting, any(), and all()

```python
# sorted()
names = ["John", "Alexander", "Bob"]
print("Sorted by length:", sorted(names, key=lambda name: len(name)))

# any()
numbers_any = [2, 4, 7, 8]
has_odd = any(n % 2 != 0 for n in numbers_any)
print("Has any odd number?", has_odd)

# all()
numbers_all = [2, 4, 6, 8]
all_even = all(n % 2 == 0 for n in numbers_all)
print("Are all numbers even?", all_even)
```

## 5. Functional Style Pipelines

Combining filter, map, and reduce to create functional style data processing pipelines.

#### Examples

```python
# Example 1: Sum of squares of even numbers
numbers = [1, 2, 3, 4, 5, 6]
sum_squares_even = reduce(
    lambda total, val: total + val,
    map(
        lambda x: x * x,
        filter(lambda x: x % 2 == 0, numbers)
    )
)
print("Sum of squares of even numbers:", sum_squares_even)

# Example 2: Average of passed students (score >= 75)
students = [
    ("John", 82),
    ("Alice", 95),
    ("Bob", 68),
    ("David", 75)
]
passed_marks = list(map(
    lambda s: s[1],
    filter(lambda s: s[1] >= 75, students)
))
average = sum(passed_marks) / len(passed_marks)
print("Average passed marks:", average)
```
