## 1. Functions are Objects

In Python, functions are first-class citizens. They are objects and can be:
- Assigned to variables
- Passed as arguments to other functions
- Returned from other functions
- Stored in collections like lists or dictionaries

### Assigning a Function

````python
def greet():
    return "Hello"

message = greet  # Assigning the function object
print(message())
````

### Storing Functions in a List

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

operations = [add, subtract]

print("Add:", operations[0](10, 5))
print("Subtract:", operations[1](10, 5))
````

## 2. Higher-Order Functions

A **higher-order function** is a function that either accepts one or more functions as arguments or returns a function.

### Passing Functions as Arguments

````python
def greet():
    print("Hello")

def execute(func):
    func()

execute(greet)
````

### Returning Functions

````python
def outer():
    def inner():
        print("Hello")
    return inner

message = outer()
message()
````

### Calculation Example

````python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def calculate(operation, a, b):
    return operation(a, b)

print("Calculate Add:", calculate(add, 10, 5))
print("Calculate Multiply:", calculate(multiply, 10, 5))
````

## 3. Closures

A **closure** is a nested function that retains reference to local variables from its enclosing scope even after the outer function has finished executing.

### Example: Normal Function (No State Preservation)

````python
def counter():
    count = 0
    count += 1
    return count

print(counter())
print(counter())
print(counter())
````

### Example: Using a Closure to Preserve State

````python
def counter():
    count = 0
    
    def increment():
        nonlocal count
        count += 1
        return count
        
    return increment

c = counter()
print(c())
print(c())
print(c())
````

## 4. Decorators

A **decorator** is a function that extends or modifies the behavior of another function without changing its source code. Decorators are applied using the `@decorator_name` syntax.

### Creating a Decorator (Manual Binding)

````python
def logger(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

def greet():
    print("Hello")

greet = logger(greet)
greet()
````

### Using `@` Syntax

````python
def logger(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@logger
def greet():
    print("Hello")

greet()
````

### Decorating Functions with Parameters (`*args`, `**kwargs`)

````python
def logger(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        result = func(*args, **kwargs)
        print("Function Completed")
        return result
    return wrapper

@logger
def multiply(a, b):
    return a * b

@logger
def greet_person(name):
    print(f"Hello {name}")

print(multiply(10, 5))
greet_person("Alice")
````
