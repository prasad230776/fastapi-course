# Python Functions - Practice
This notebook covers defining functions, parameters & arguments, scopes, first-class functions, lambdas, generators, closures, decorators, and higher-order functions.

### 1. Defining and Calling Functions
A function is defined using the `def` keyword followed by a name, parentheses, and a colon.

```python
def greet():
    print("Hello, world!")
    print("Welcome to Python!")

greet()
```

### 2. Parameters & Arguments
By default, Python matches arguments to parameters by position.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

describe_pet("hamster", "Harry")
```

### Concept: Default Values
Set default values for parameters to make them optional.

```python
def greet_user(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet_user("Alice")
greet_user("Bob", "Hi")
```

### Concept: Keyword Arguments
Call functions using parameter names to pass them in any order.

```python
def create_profile(name, age, city):
    print(f"{name}, {age}, from {city}")

create_profile(city="New York", name="Alice", age=25)
```

### 3. Return Values
### Concept: Returning Single and Multiple Values
Use `return` to send values back. Separate multiple outputs with commas to return them as a tuple.

```python
def get_min_max(numbers):
    return min(numbers), max(numbers)

minimum, maximum = get_min_max([5, 2, 8, 1, 9])
print(f"Min: {minimum}, Max: {maximum}")
```

### 4. Flexible & Enforced Arguments
### Concept: *args and **kwargs
`*args` collects additional positional arguments into a tuple. `**kwargs` collects keyword arguments into a dictionary.

```python
def display(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

display(1, 2, 3, name="Alice", age=20)
```

### Concept: Positional-Only and Keyword-Only Parameters
Use `/` to mark positional-only parameters and `*` to mark keyword-only parameters.

```python
def mix_example(first, second, /, *, third):
    print(first, second, third)

mix_example(1, 2, third=3)
```

### 5. Variable Scope & LEGB Rule
### Concept: LEGB Scope Resolution
Python searches variables in the order: Local, Enclosing, Global, and Built-in.

```python
x = "global"
def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("Inner x:", x)
    inner()

outer()
```

### Concept: global and nonlocal Keywords
Use `global` to modify global variables and `nonlocal` to modify enclosing (outer) variables.

```python
counter = 0
def increment_global():
    global counter
    counter += 1

increment_global()
print("Global Counter:", counter)
```

### 6. First-Class Functions
### Concept: Assigning and Passing Functions
Functions are objects and can be assigned to variables or passed as arguments.

```python
def greet(name):
    return f"Hello, {name}!"

say_hello = greet
print(say_hello("Alice"))
```

### 7. Lambda Functions
### Concept: Single-Expression Anonymous Functions
Use lambda for simple inline operations without full function definitions.

```python
add_lambda = lambda x, y: x + y
print("Lambda addition:", add_lambda(3, 7))
```

### 8. Generators
### Concept: Creating a Generator
Generators use the `yield` keyword to return values lazily one-at-a-time, optimizing memory usage.

```python
def count_up_to(n):
    for i in range(1, n + 1):
        yield i

numbers = count_up_to(5)
for num in numbers:
    print(num)
```

### Concept: Manually Iterating with next()
Retrieve individual values from a generator manually using the `next()` function.

```python
def countdown():
    yield 3
    yield 2
    yield 1

gen = countdown()
print(next(gen))
print(next(gen))
print(next(gen))
```

### Concept: Generator Expressions
Write inline generator expressions using parenthesis `()` similar to list comprehensions.

```python
squares = (x ** 2 for x in range(5))
for square in squares:
    print(square)
```

### 9. Closures & Decorators
### Concept: Closures
Closures retain access to their enclosing scope variables even after execution finishes.

```python
def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = make_multiplier(2)
print(double(10))
```

### Concept: Basic Decorator
Decorators wrap function calls to modify execution workflows dynamically.

```python
def my_decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@my_decorator
def greet_msg():
    print("Hello")

greet_msg()
```

### Concept: Decorator with Arguments
Implement wrappers utilizing `*args` and `**kwargs` to intercept calls with arbitrary arguments.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(10, 20))
```

### 10. Declarative Programming & Higher-Order Functions
### Concept: Imperative vs Declarative
Contrast loop-based code with functional `map` pipelines.

```python
numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
```

### Concept: map()
Apply a function to all elements in an iterable.

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)
```

### Concept: filter()
Filter elements based on a truth condition.

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

### Concept: reduce()
Accumulate elements using a function into a single value.

```python
from functools import reduce
numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)
print(product)
```

### Concept: Pythonic Alternatives (Comprehensions)

Functional Pipeline Example

The following example combines `filter()`, `map()`, and `reduce()` to compute the **sum of squares of even numbers**.

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

evens = filter(lambda x: x % 2 == 0, numbers)
squares = map(lambda x: x**2, evens)

total = reduce(lambda x, y: x + y, squares)

print(total)
```

Use generator expressions or comprehensions instead of maps and filters.

```python
numbers = [1, 2, 3, 4, 5, 6]
total = sum(x**2 for x in numbers if x % 2 == 0)
print(total)
```
